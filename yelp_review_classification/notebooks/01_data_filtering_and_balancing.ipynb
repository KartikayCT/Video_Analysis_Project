```python
import pandas as pd
import numpy as np
import os
```

```python
train_path = "../data/train.csv"
test_path = "../data/test.csv"
```

```python
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Train dataset shape:", train_df.shape)
print("Test dataset shape:", test_df.shape)
```

```python
train_df.head()
```

```python
test_df.head()
```

```python
print("Train columns:")
print(train_df.columns.tolist())

print("\nTest columns:")
print(test_df.columns.tolist())
```

```python
print(train_df.dtypes)
print()
print(test_df.dtypes)
```

```python
print("Missing values in train dataset:")
print(train_df.isnull().sum())

print("\nMissing values in test dataset:")
print(test_df.isnull().sum())
```

```python
print("Train class distribution:")
print(train_df["class_index"].value_counts().sort_index())

print("\nTest class distribution:")
print(test_df["class_index"].value_counts().sort_index())
```

```python
df = pd.concat(
    [train_df, test_df],
    ignore_index=True
)

print("Combined dataset shape:", df.shape)
```

```python
class_distribution = (
    df["class_index"]
    .value_counts()
    .sort_index()
)

print(class_distribution)
```

```python
df_filtered = df[
    df["class_index"].isin([1, 3, 5])
].copy()

print("Shape after filtering:", df_filtered.shape)
```

```python
sentiment_mapping = {
    1: "Negative",
    3: "Neutral",
    5: "Positive"
}

df_filtered["sentiment"] = df_filtered["class_index"].map(
    sentiment_mapping
)

df_filtered.head()
```

```python
print(
    df_filtered["sentiment"]
    .value_counts()
)
```

```python
random_state = 42

negative_samples = df_filtered[
    df_filtered["sentiment"] == "Negative"
].sample(
    n=50000,
    random_state=random_state
)

neutral_samples = df_filtered[
    df_filtered["sentiment"] == "Neutral"
].sample(
    n=50000,
    random_state=random_state
)

positive_samples = df_filtered[
    df_filtered["sentiment"] == "Positive"
].sample(
    n=50000,
    random_state=random_state
)
```

```python
balanced_df = pd.concat(
    [
        negative_samples,
        neutral_samples,
        positive_samples
    ],
    ignore_index=True
)
```

```python
balanced_df = balanced_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
print("Final dataset shape:", balanced_df.shape)
```

```python
print(
    balanced_df["sentiment"]
    .value_counts()
)
```

```python
print(
    balanced_df.isnull().sum()
)
```

```python
os.makedirs("../data/processed", exist_ok=True)
output_path = "../data/processed/yelp_balanced_150k.csv"

balanced_df.to_csv(
    output_path,
    index=False
)

print("Balanced dataset saved successfully.")
print("Location:", output_path)
```
